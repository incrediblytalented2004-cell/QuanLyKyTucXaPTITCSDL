```python
import os
import xml.etree.ElementTree as ET
from collections import Counter

path = "/content/UAV_dataset/dataset_xml_format"

xml_files = [
    f for f in os.listdir(path)
    if f.lower().endswith(".xml")
]

class_counter = Counter()
object_counts = []
invalid_boxes = 0
error_xml = 0

for xml_file in xml_files:
    xml_path = os.path.join(path, xml_file)

    try:
        root = ET.parse(xml_path).getroot()

        objects = root.findall("object")
        object_counts.append(len(objects))

        for obj in objects:
            name = obj.findtext("name", default="UNKNOWN").strip()
            class_counter[name] += 1

            box = obj.find("bndbox")

            if box is not None:
                xmin = float(box.findtext("xmin"))
                ymin = float(box.findtext("ymin"))
                xmax = float(box.findtext("xmax"))
                ymax = float(box.findtext("ymax"))

                if xmin >= xmax or ymin >= ymax:
                    invalid_boxes += 1

    except Exception:
        error_xml += 1

print("Số XML kiểm tra:", len(xml_files))
print("Số XML lỗi:", error_xml)

print("\nCác class:")
for name, count in class_counter.items():
    print(f"  {name}: {count}")

print("\nSố object trung bình / ảnh:",
      sum(object_counts) / len(object_counts))

print("Bounding box không hợp lệ:", invalid_boxes)
```

```python
from torch.utils.data import DataLoader

def collate_fn(batch):
    images = []
    boxes = []

    for image, target_boxes in batch:
        images.append(image)
        boxes.append(target_boxes)

    images = torch.stack(images)

    return images, boxes


train_loader = DataLoader(
    train_dataset,
    batch_size=16,
    shuffle=True,
    num_workers=2,
    collate_fn=collate_fn
)

val_loader = DataLoader(
    val_dataset,
    batch_size=16,
    shuffle=False,
    num_workers=2,
    collate_fn=collate_fn
)

print("Train batches:", len(train_loader))
print("Validation batches:", len(val_loader))
```

```python
import torch
import torch.nn as nn


class UAVCNN(nn.Module):

    def __init__(self):
        super().__init__()

        self.features = nn.Sequential(

            # Block 1
            nn.Conv2d(3, 16, kernel_size=3, padding=1),
            nn.ReLU(inplace=True),
            nn.MaxPool2d(2),

            # Block 2
            nn.Conv2d(16, 32, kernel_size=3, padding=1),
            nn.ReLU(inplace=True),
            nn.MaxPool2d(2),

            # Block 3
            nn.Conv2d(32, 64, kernel_size=3, padding=1),
            nn.ReLU(inplace=True),
            nn.MaxPool2d(2),

            # Block 4
            nn.Conv2d(64, 128, kernel_size=3, padding=1),
            nn.ReLU(inplace=True),
            nn.MaxPool2d(2),

            # Block 5
            nn.Conv2d(128, 128, kernel_size=3, padding=1),
            nn.ReLU(inplace=True)
        )

        # 5 đầu ra:
        # objectness + xmin + ymin + xmax + ymax
        self.detection_head = nn.Conv2d(
            128,
            5,
            kernel_size=1
        )

    def forward(self, x):

        x = self.features(x)

        x = self.detection_head(x)

        return x
```

```python
import torch
import torch.nn as nn


GRID_SIZE = 20


def create_targets(boxes_batch, grid_size=GRID_SIZE):
    """
    Chuyển bounding box thành target cho lưới grid.

    Mỗi ô:
    [objectness, x_center, y_center, width, height]

    Tọa độ đều được chuẩn hóa về 0 -> 1.
    """

    batch_size = len(boxes_batch)

    targets = torch.zeros(
        batch_size,
        5,
        grid_size,
        grid_size
    )

    for b, boxes in enumerate(boxes_batch):

        for box in boxes:

            xmin, ymin, xmax, ymax = box

            # Tâm bounding box
            x_center = (xmin + xmax) / 2
            y_center = (ymin + ymax) / 2

            # Kích thước bounding box
            width = xmax - xmin
            height = ymax - ymin

            # Xác định ô grid
            grid_x = min(
                int(x_center * grid_size),
                grid_size - 1
            )

            grid_y = min(
                int(y_center * grid_size),
                grid_size - 1
            )

            # Ghi target
            targets[b, 0, grid_y, grid_x] = 1.0

            targets[b, 1, grid_y, grid_x] = x_center
            targets[b, 2, grid_y, grid_x] = y_center
            targets[b, 3, grid_y, grid_x] = width
            targets[b, 4, grid_y, grid_x] = height

    return targets
```

```python
class UAVLoss(nn.Module):

    def __init__(self):
        super().__init__()

        self.objectness_loss = nn.BCEWithLogitsLoss()
        self.box_loss = nn.MSELoss()

    def forward(self, predictions, targets):

        # Objectness
        pred_obj = predictions[:, 0]
        target_obj = targets[:, 0]

        obj_loss = self.objectness_loss(
            pred_obj,
            target_obj
        )

        # Bounding box
        pred_box = predictions[:, 1:5]
        target_box = targets[:, 1:5]

        # Chỉ tính box loss tại những ô có UAV
        mask = target_obj > 0

        if mask.any():

            pred_box = pred_box.permute(0, 2, 3, 1)[mask]
            target_box = target_box.permute(0, 2, 3, 1)[mask]

            box_loss = self.box_loss(
                pred_box,
                target_box
            )

        else:
            box_loss = torch.tensor(
                0.0,
                device=predictions.device
            )

        total_loss = obj_loss + 5.0 * box_loss

        return total_loss, obj_loss, box_loss
```

```python
images, boxes = next(iter(train_loader))

images = images.to(device)

targets = create_targets(boxes).to(device)

with torch.no_grad():
    predictions = model(images)

criterion = UAVLoss()

loss, obj_loss, box_loss = criterion(
    predictions,
    targets
)

print("Prediction shape:", predictions.shape)
print("Target shape:", targets.shape)

print("Total loss:", loss.item())
print("Objectness loss:", obj_loss.item())
print("Box loss:", box_loss.item())
```

```python
import torch.optim as optim

optimizer = optim.Adam(
    model.parameters(),
    lr=0.001
)

num_epochs = 30

train_losses = []
val_losses = []

for epoch in range(num_epochs):

    # =========================
    # TRAIN
    # =========================

    model.train()

    running_train_loss = 0.0

    for images, boxes in train_loader:

        images = images.to(device)

        targets = create_targets(boxes).to(device)

        optimizer.zero_grad()

        predictions = model(images)

        loss, obj_loss, box_loss = criterion(
            predictions,
            targets
        )

        loss.backward()

        optimizer.step()

        running_train_loss += loss.item()

    train_loss = (
        running_train_loss /
        len(train_loader)
    )

    # =========================
    # VALIDATION
    # =========================

    model.eval()

    running_val_loss = 0.0

    with torch.no_grad():

        for images, boxes in val_loader:

            images = images.to(device)

            targets = create_targets(boxes).to(device)

            predictions = model(images)

            loss, _, _ = criterion(
                predictions,
                targets
            )

            running_val_loss += loss.item()

    val_loss = (
        running_val_loss /
        len(val_loader)
    )

    train_losses.append(train_loss)
    val_losses.append(val_loss)

    print(
        f"Epoch [{epoch + 1:02d}/{num_epochs}] "
        f"Train Loss: {train_loss:.4f} "
        f"Val Loss: {val_loss:.4f}"
    )
```

```python
# ==========================================
# STRUCTURED PRUNING 25% - FILTER PRUNING
# ==========================================

import torch
import torch.nn as nn


# ------------------------------------------
# 1. Xác định số channel sau pruning
# ------------------------------------------

keep_channels = {
    "conv1": 12,   # 16 -> 12
    "conv2": 24,   # 32 -> 24
    "conv3": 48,   # 64 -> 48
    "conv4": 96,   # 128 -> 96
    "conv5": 96    # 128 -> 96
}


# ------------------------------------------
# 2. Tạo model sau pruning
# ------------------------------------------

class UAVCNN_Pruned(nn.Module):

    def __init__(self):
        super().__init__()

        self.features = nn.Sequential(

            # Conv1
            nn.Conv2d(3, 12, 3, padding=1),
            nn.ReLU(inplace=True),
            nn.MaxPool2d(2),

            # Conv2
            nn.Conv2d(12, 24, 3, padding=1),
            nn.ReLU(inplace=True),
            nn.MaxPool2d(2),

            # Conv3
            nn.Conv2d(24, 48, 3, padding=1),
            nn.ReLU(inplace=True),
            nn.MaxPool2d(2),

            # Conv4
            nn.Conv2d(48, 96, 3, padding=1),
            nn.ReLU(inplace=True),
            nn.MaxPool2d(2),

            # Conv5
            nn.Conv2d(96, 96, 3, padding=1),
            nn.ReLU(inplace=True)
        )

        # Không pruning detection head
        self.detection_head = nn.Conv2d(96, 5, kernel_size=1)


    def forward(self, x):
        x = self.features(x)
        x = self.detection_head(x)
        return x


# ------------------------------------------
# 3. Tạo model mới
# ------------------------------------------

pruned_model = UAVCNN_Pruned().to(device)


# ------------------------------------------
# 4. Lấy các Conv của model Dense
# ------------------------------------------

dense_convs = [
    model.features[0],
    model.features[3],
    model.features[6],
    model.features[9],
    model.features[12]
]

pruned_convs = [
    pruned_model.features[0],
    pruned_model.features[3],
    pruned_model.features[6],
    pruned_model.features[9],
    pruned_model.features[12]
]


# ------------------------------------------
# 5. Copy các filter quan trọng nhất
# ------------------------------------------

previous_indices = None

for dense_conv, pruned_conv in zip(dense_convs, pruned_convs):

    weights = dense_conv.weight.data

    # L1-norm của từng output filter
    importance = torch.sum(
        torch.abs(weights),
        dim=(1, 2, 3)
    )

    # Lấy các filter quan trọng nhất
    num_keep = pruned_conv.out_channels

    _, indices = torch.topk(
        importance,
        num_keep
    )

    # Sắp xếp index để giữ thứ tự ổn định
    indices = torch.sort(indices).values

    # Nếu layer trước đã bị pruning,
    # chỉ lấy các input channel tương ứng
    if previous_indices is not None:
        weights = weights[:, previous_indices, :, :]

    # Copy weight
    pruned_conv.weight.data.copy_(
        weights[indices, :, :, :]
    )

    # Copy bias nếu có
    if dense_conv.bias is not None:
        pruned_conv.bias.data.copy_(
            dense_conv.bias.data[indices]
        )

    # Lưu index để dùng cho layer tiếp theo
    previous_indices = indices


# ------------------------------------------
# 6. Copy detection head
# ------------------------------------------

dense_head = model.detection_head
pruned_head = pruned_model.detection_head

head_weights = dense_head.weight.data

# Chỉ lấy các input channels còn lại
head_weights = head_weights[:, previous_indices, :, :]

pruned_head.weight.data.copy_(head_weights)

if dense_head.bias is not None:
    pruned_head.bias.data.copy_(dense_head.bias.data)


# ------------------------------------------
# 7. Kiểm tra model
# ------------------------------------------

print("===== MODEL SAU STRUCTURED PRUNING =====")

print(pruned_model)

print("\nKiểm tra output:")

test_input = torch.randn(1, 3, 320, 320).to(device)

with torch.no_grad():
    output = pruned_model(test_input)

print("Input :", test_input.shape)
print("Output:", output.shape)


# ------------------------------------------
# 8. Tính parameters
# ------------------------------------------

total_params_pruned = sum(
    p.numel()
    for p in pruned_model.parameters()
)

total_params_dense = sum(
    p.numel()
    for p in model.parameters()
)

print("\n===== PARAMETERS =====")

print(f"Dense  : {total_params_dense:,}")
print(f"Pruned : {total_params_pruned:,}")

reduction_params = (
    1 - total_params_pruned / total_params_dense
) * 100

print(
    f"Giảm parameters: {reduction_params:.2f}%"
)
```

```python
import torch
import torch.nn as nn

# ============================================================
# 1. DENSE CNN
# ============================================================

class UAVCNN(nn.Module):
    def __init__(self):
        super().__init__()

        self.features = nn.Sequential(
            nn.Conv2d(3, 16, 3, padding=1),
            nn.ReLU(inplace=True),
            nn.MaxPool2d(2),

            nn.Conv2d(16, 32, 3, padding=1),
            nn.ReLU(inplace=True),
            nn.MaxPool2d(2),

            nn.Conv2d(32, 64, 3, padding=1),
            nn.ReLU(inplace=True),
            nn.MaxPool2d(2),

            nn.Conv2d(64, 128, 3, padding=1),
            nn.ReLU(inplace=True),
            nn.MaxPool2d(2),

            nn.Conv2d(128, 128, 3, padding=1),
            nn.ReLU(inplace=True)
        )

        self.detection_head = nn.Conv2d(
            128, 5, kernel_size=1
        )

    def forward(self, x):
        x = self.features(x)
        x = self.detection_head(x)
        return x


# ============================================================
# 2. PRUNED CNN
# ============================================================

class UAVCNN_Pruned(nn.Module):
    def __init__(self):
        super().__init__()

        self.features = nn.Sequential(
            nn.Conv2d(3, 12, 3, padding=1),
            nn.ReLU(inplace=True),
            nn.MaxPool2d(2),

            nn.Conv2d(12, 24, 3, padding=1),
            nn.ReLU(inplace=True),
            nn.MaxPool2d(2),

            nn.Conv2d(24, 48, 3, padding=1),
            nn.ReLU(inplace=True),
            nn.MaxPool2d(2),

            nn.Conv2d(48, 96, 3, padding=1),
            nn.ReLU(inplace=True),
            nn.MaxPool2d(2),

            nn.Conv2d(96, 96, 3, padding=1),
            nn.ReLU(inplace=True)
        )

        self.detection_head = nn.Conv2d(
            96, 5, kernel_size=1
        )

    def forward(self, x):
        x = self.features(x)
        x = self.detection_head(x)
        return x


# ============================================================
# 3. LOAD DENSE MODEL
# ============================================================

dense_model = UAVCNN()

dense_checkpoint = torch.load(
    "/content/UAV_dataset/uav_cnn_dense_baseline.pth",
    map_location="cpu"
)

dense_model.load_state_dict(dense_checkpoint)
dense_model.eval()


# ============================================================
# 4. LOAD PRUNED + FINE-TUNED MODEL
# ============================================================

pruned_model = UAVCNN_Pruned()

pruned_checkpoint = torch.load(
    "/content/UAV_dataset/uav_cnn_pruned25_finetuned.pth",
    map_location="cpu"
)

pruned_model.load_state_dict(pruned_checkpoint)
pruned_model.eval()


# ============================================================
# 5. HÀM TÍNH IoU ĐÚNG
# ============================================================

def calculate_mean_iou(model, dataloader):

    model.eval()

    total_iou = 0.0
    count = 0

    with torch.no_grad():

        for images, gt_boxes in dataloader:

            predictions = model(images)

            # Objectness
            obj = torch.sigmoid(predictions[:, 0])

            for b in range(images.shape[0]):

                # Grid cell có objectness cao nhất
                idx = torch.argmax(obj[b])

                gy = idx // GRID_SIZE
                gx = idx % GRID_SIZE

                # ==========================================
                # MODEL OUTPUT:
                # [x_center, y_center, width, height]
                # ==========================================

                pred = predictions[b, 1:5, gy, gx]

                cx, cy, w, h = pred

                # Decode:
                # center + width/height
                # -> xmin, ymin, xmax, ymax

                px1 = cx - w / 2
                py1 = cy - h / 2
                px2 = cx + w / 2
                py2 = cy + h / 2

                # Clamp
                px1 = torch.clamp(px1, 0, 1).item()
                py1 = torch.clamp(py1, 0, 1).item()
                px2 = torch.clamp(px2, 0, 1).item()
                py2 = torch.clamp(py2, 0, 1).item()

                # ==========================================
                # GROUND TRUTH:
                # [xmin, ymin, xmax, ymax]
                # ==========================================

                gt = gt_boxes[b][0]

                gx1, gy1, gx2, gy2 = gt.tolist()

                # ==========================================
                # INTERSECTION
                # ==========================================

                ix1 = max(px1, gx1)
                iy1 = max(py1, gy1)
                ix2 = min(px2, gx2)
                iy2 = min(py2, gy2)

                inter_w = max(0, ix2 - ix1)
                inter_h = max(0, iy2 - iy1)

                inter_area = inter_w * inter_h

                # ==========================================
                # AREA
                # ==========================================

                pred_area = (
                    max(0, px2 - px1) *
                    max(0, py2 - py1)
                )

                gt_area = (
                    max(0, gx2 - gx1) *
                    max(0, gy2 - gy1)
                )

                union_area = (
                    pred_area +
                    gt_area -
                    inter_area
                )

                if union_area > 0:
                    iou = inter_area / union_area
                else:
                    iou = 0

                total_iou += iou
                count += 1

    return total_iou / count


# ============================================================
# 6. TÍNH MEAN IoU
# ============================================================

print("============================================")
print("     ĐÁNH GIÁ MEAN IoU - DECODER ĐÚNG")
print("============================================")

dense_iou = calculate_mean_iou(
    dense_model,
    val_loader
)

pruned_iou = calculate_mean_iou(
    pruned_model,
    val_loader
)

mixed_iou = calculate_mean_iou(
    mixed_model,
    val_loader
)


# ============================================================
# 7. IN KẾT QUẢ
# ============================================================

print()
print("===== KẾT QUẢ =====")

print(f"Dense FP32              : {dense_iou:.6f}")
print(f"Pruned 25% + FT FP32    : {pruned_iou:.6f}")
print(f"Mixed INT8 + FP32 Head  : {mixed_iou:.6f}")

print()
print("============================================")
print("                 SO SÁNH")
print("============================================")

print(f"{'Model':<30} {'Parameters':>12} {'MAC':>15} {'Mean IoU':>12}")
print("-" * 75)

print(
    f"{'Dense FP32':<30} "
    f"{245669:>12,} "
    f"{'158.77 MMAC':>15} "
    f"{dense_iou:>12.6f}"
)

print(
    f"{'Pruned 25% + FT FP32':<30} "
    f"{138461:>12,} "
    f"{'91.43 MMAC':>15} "
    f"{pruned_iou:>12.6f}"
)

print(
    f"{'Mixed INT8 + FP32 Head':<30} "
    f"{138461:>12,} "
    f"{'91.43 MMAC':>15} "
    f"{mixed_iou:>12.6f}"
)

print()
print("Hoàn tất đánh giá.")
```

```python

import torch
import torch.nn as nn
import torch.ao.nn.quantized as nnq
import torch.ao.quantization as quant

torch.backends.quantized.engine = "fbgemm"

path = "/content/drive/MyDrive/UAV_Project/uav_cnn_pruned25_mixed_int8_fp32head.pth"
ckpt = torch.load(path, map_location="cpu", weights_only=True)

class UAVCNN_Pruned_Mixed_INT8(nn.Module):
    def __init__(self):
        super().__init__()
        self.quant = nnq.Quantize(scale=1.0, zero_point=0, dtype=torch.quint8)

        self.features = nn.Sequential(
            nnq.Conv2d(3, 12, 3, padding=1),
            nn.ReLU(inplace=True), nn.MaxPool2d(2),
            nnq.Conv2d(12, 24, 3, padding=1),
            nn.ReLU(inplace=True), nn.MaxPool2d(2),
            nnq.Conv2d(24, 48, 3, padding=1),
            nn.ReLU(inplace=True), nn.MaxPool2d(2),
            nnq.Conv2d(48, 96, 3, padding=1),
            nn.ReLU(inplace=True), nn.MaxPool2d(2),
            nnq.Conv2d(96, 96, 3, padding=1),
            nn.ReLU(inplace=True)
        )

        self.dequant = nnq.DeQuantize()
        self.detection_head = nn.Conv2d(96, 5, kernel_size=1)

    def forward(self, x):
        x = self.quant(x)
        x = self.features(x)
        x = self.dequant(x)
        return self.detection_head(x)

model_test = UAVCNN_Pruned_Mixed_INT8().eval()

# Khôi phục scale và zero_point của từng lớp
for name, module in model_test.named_modules():
    if isinstance(module, nnq.Conv2d):
        module.scale = float(ckpt[f"{name}.scale"].item())
        module.zero_point = int(ckpt[f"{name}.zero_point"].item())

try:
    result = model_test.load_state_dict(ckpt, strict=True)
    print("Nạp checkpoint thành công:", result)
except Exception as e:
    print("Lỗi nạp checkpoint:", type(e).__name__, str(e)[:1500])
```

```python

import rarfile, io, os, torch
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
import torchvision.transforms as transforms
from matplotlib.patches import Rectangle

rar_path = "/content/drive/MyDrive/UAV_Project/UAV_train.rar"
rf = rarfile.RarFile(rar_path)
names = rf.namelist()

# Ghép ảnh với nhãn TXT theo tên file
label_files = [
    n for n in names
    if "/labels/" in n.lower() and n.lower().endswith(".txt")
]

sample = None
for label_name in label_files:
    base = os.path.splitext(os.path.basename(label_name))[0]
    split = label_name.split("/labels/")[1].split("/")[0].lower()
    candidates = [
        n for n in names
        if f"/images/{split}/" in n.lower()
        and os.path.splitext(os.path.basename(n))[0] == base
        and n.lower().endswith((".jpg", ".jpeg", ".png"))
    ]
    if candidates:
        sample = (candidates[0], label_name)
        break

if sample is None:
    raise RuntimeError("Không tìm được cặp ảnh và nhãn tương ứng.")

image_name, label_name = sample
image = Image.open(io.BytesIO(rf.read(image_name))).convert("RGB")
original_w, original_h = image.size

transform = transforms.Compose([
    transforms.Resize((320, 320)),
    transforms.ToTensor()
])
x = transform(image).unsqueeze(0)

model_test.eval()
with torch.no_grad():
    pred = model_test(x)[0]

# Lấy ô có objectness cao nhất
obj = torch.sigmoid(pred[0])
idx = int(obj.argmax())
gy, gx = divmod(idx, obj.shape[1])
score = obj[gy, gx].item()

# Giải mã bbox giống notebook cũ
box = torch.sigmoid(pred[1:5, gy, gx])
xc, yc, bw, bh = [v.item() for v in box]

# Quy đổi tọa độ chuẩn hóa về ảnh gốc
xmin = max(0, min(original_w, (xc - bw / 2) * original_w))
ymin = max(0, min(original_h, (yc - bh / 2) * original_h))
xmax = max(0, min(original_w, (xc + bw / 2) * original_w))
ymax = max(0, min(original_h, (yc + bh / 2) * original_h))

fig, ax = plt.subplots(figsize=(10, 6))
ax.imshow(image)
ax.add_patch(Rectangle(
    (xmin, ymin), xmax - xmin, ymax - ymin,
    fill=False, edgecolor="red", linewidth=2
))
ax.set_title(f"UAV prediction | score={score:.3f}")
ax.axis("off")
plt.show()

print("Ảnh:", image_name)
print("Nhãn TXT:", label_name)
print("Objectness:", round(score, 4))
print("BBox dự đoán (x1,y1,x2,y2):",
      tuple(round(v, 1) for v in (xmin, ymin, xmax, ymax)))
print("Nhãn thật:", rf.read(label_name).decode("utf-8", errors="replace"))
```

```python

# ============================================================
# KIEM TRA DATASET VA DANH GIA CNN MIXED INT8
# KHONG HUAN LUYEN LAI
# ============================================================

import os
import random
import shutil
import subprocess
import xml.etree.ElementTree as ET
from pathlib import Path

import torch
import torch.nn as nn
import torch.ao.nn.quantized as nnq
from torch.utils.data import Dataset, DataLoader
from PIL import Image
import torchvision.transforms as transforms

from google.colab import drive
drive.mount("/content/drive")

# 1. Duong dan
project = Path("/content/drive/MyDrive/UAV_Project")
archive = project / "dataset_xml_format.rar"
checkpoint = project / "uav_cnn_pruned25_mixed_int8_fp32head.pth"

dataset_dir = Path("/content/UAV_dataset/dataset_xml_format")
split_dir = Path("/content/UAV_dataset/split")

assert checkpoint.exists(), f"Khong tim thay checkpoint: {checkpoint}"
assert archive.exists(), f"Khong tim thay dataset XML: {archive}"

# 2. Khoi phuc dataset XML neu Colab mat du lieu
if not dataset_dir.exists():
    if shutil.which("unrar") is None:
        subprocess.run(
            ["apt-get", "-qq", "update"], check=True
        )
        subprocess.run(
            ["apt-get", "-qq", "install", "-y", "unrar"],
            check=True
        )

    subprocess.run([
        "unrar", "x", "-o-", str(archive),
        "/content/UAV_dataset/"
    ], check=True)

assert dataset_dir.exists(), "Giai nen dataset that bai"

# 3. Kiem tra cap anh/XML
images = sorted([
    p for p in dataset_dir.iterdir()
    if p.is_file() and p.suffix.lower() in
    [".jpg", ".jpeg", ".png"]
])

pairs = [
    p for p in images
    if p.with_suffix(".xml").exists()
]

print("So anh:", len(images))
print("So cap anh/XML:", len(pairs))
print("Checkpoint:", round(checkpoint.stat().st_size / 1024, 2), "KB")

assert len(pairs) > 0, "Khong co cap anh/XML hop le"

# 4. Tao lai split neu chua co hoac khong du 877/220 anh
train_img = split_dir / "train/images"
train_xml = split_dir / "train/annotations"
val_img = split_dir / "val/images"
val_xml = split_dir / "val/annotations"

def count_images(folder):
    if not folder.exists():
        return 0
    return sum(
        p.suffix.lower() in [".jpg", ".jpeg", ".png"]
        for p in folder.iterdir()
        if p.is_file()
    )

split_ok = (
    count_images(train_img) == int(len(pairs) * 0.8)
    and count_images(val_img) == len(pairs) - int(len(pairs) * 0.8)
    and count_images(train_xml) == count_images(train_img)
    and count_images(val_xml) == count_images(val_img)
)

if not split_ok:
    if split_dir.exists():
        shutil.rmtree(split_dir)

    for folder in [train_img, train_xml, val_img, val_xml]:
        folder.mkdir(parents=True, exist_ok=True)

    items = pairs.copy()
    random.Random(42).shuffle(items)
    cut = int(len(items) * 0.8)

    for subset, subset_img, subset_xml in [
        (items[:cut], train_img, train_xml),
        (items[cut:], val_img, val_xml)
    ]:
        for img in subset:
            shutil.copy2(img, subset_img / img.name)
            shutil.copy2(
                img.with_suffix(".xml"),
                subset_xml / img.with_suffix(".xml").name
            )

print("Train:", count_images(train_img))
print("Validation:", count_images(val_img))

# 5. Dataset XML: chi lay object drone
class UAVValidationDataset(Dataset):
    def __init__(self, image_dir, annotation_dir):
        self.image_dir = Path(image_dir)
        self.annotation_dir = Path(annotation_dir)
        self.images = sorted([
            p for p in self.image_dir.iterdir()
            if p.suffix.lower() in [".jpg", ".jpeg", ".png"]
            and (self.annotation_dir / (p.stem + ".xml")).exists()
        ])
        self.transform = transforms.Compose([
            transforms.Resize((320, 320)),
            transforms.ToTensor()
        ])

    def __len__(self):
        return len(self.images)

    def __getitem__(self, index):
        path = self.images[index]
        image = Image.open(path).convert("RGB")
        root = ET.parse(
            self.annotation_dir / (path.stem + ".xml")
        ).getroot()

        size = root.find("size")
        ow = float(size.findtext("width"))
        oh = float(size.findtext("height"))

        boxes = []
        for obj in root.findall("object"):
            if obj.findtext("name") != "drone":
                continue
            b = obj.find("bndbox")
            boxes.append([
                float(b.findtext("xmin")) / ow,
                float(b.findtext("ymin")) / oh,
                float(b.findtext("xmax")) / ow,
                float(b.findtext("ymax")) / oh
            ])

        return self.transform(image), torch.tensor(
            boxes, dtype=torch.float32
        ), path.name

val_ds = UAVValidationDataset(val_img, val_xml)
val_loader = DataLoader(val_ds, batch_size=1, shuffle=False)

assert len(val_ds) > 0, "Validation khong co anh hop le"

# 6. Dung lai dung kien truc Mixed INT8 da luu
torch.backends.quantized.engine = "fbgemm"
ckpt = torch.load(
    checkpoint, map_location="cpu", weights_only=True
)

class UAVCNN_Pruned_Mixed_INT8(nn.Module):
    def __init__(self):
        super().__init__()
        self.quant = nnq.Quantize(
            scale=1.0, zero_point=0, dtype=torch.quint8
        )
        self.features = nn.Sequential(
            nnq.Conv2d(3, 12, 3, padding=1),
            nn.ReLU(inplace=True), nn.MaxPool2d(2),
            nnq.Conv2d(12, 24, 3, padding=1),
            nn.ReLU(inplace=True), nn.MaxPool2d(2),
            nnq.Conv2d(24, 48, 3, padding=1),
            nn.ReLU(inplace=True), nn.MaxPool2d(2),
            nnq.Conv2d(48, 96, 3, padding=1),
            nn.ReLU(inplace=True), nn.MaxPool2d(2),
            nnq.Conv2d(96, 96, 3, padding=1),
            nn.ReLU(inplace=True)
        )
        self.dequant = nnq.DeQuantize()
        self.detection_head = nn.Conv2d(96, 5, 1)

    def forward(self, x):
        x = self.quant(x)
        x = self.features(x)
        x = self.dequant(x)
        return self.detection_head(x)

model = UAVCNN_Pruned_Mixed_INT8().eval()

for name, module in model.named_modules():
    if isinstance(module, nnq.Conv2d):
        module.scale = float(ckpt[f"{name}.scale"].item())
        module.zero_point = int(ckpt[f"{name}.zero_point"].item())

model.load_state_dict(ckpt, strict=True)
print("Da nap checkpoint Mixed INT8.")

# 7. Danh gia so bo: cell objectness cao nhat va IoU
def iou_xyxy(a, b):
    ix1, iy1 = max(a[0], b[0]), max(a[1], b[1])
    ix2, iy2 = min(a[2], b[2]), min(a[3], b[3])
    inter = max(0, ix2-ix1) * max(0, iy2-iy1)
    area_a = max(0, a[2]-a[0]) * max(0, a[3]-a[1])
    area_b = max(0, b[2]-b[0]) * max(0, b[3]-b[1])
    union = area_a + area_b - inter
    return inter / union if union > 0 else 0.0

ious, scores, evaluated = [], [], 0

with torch.inference_mode():
    for image, gt_boxes, filename in val_loader:
        out = model(image)
        if not torch.isfinite(out).all():
            continue

        obj = torch.sigmoid(out[0, 0])
        idx = int(obj.argmax())
        gy, gx = divmod(idx, obj.shape[1])
        score = float(obj[gy, gx])
        raw = out[0, 1:5, gy, gx].tolist()

        # Model duoc huan luyen voi cx, cy, w, h
        cx, cy, bw, bh = raw
        pred = [
            max(0.0, min(1.0, cx-bw/2)),
            max(0.0, min(1.0, cy-bh/2)),
            max(0.0, min(1.0, cx+bw/2)),
            max(0.0, min(1.0, cy+bh/2))
        ]

        if len(gt_boxes[0]) > 0:
            # So sanh voi box drone dau tien trong anh
            gt = gt_boxes[0][0].tolist()
            ious.append(iou_xyxy(pred, gt))
            scores.append(score)
            evaluated += 1

print("\n===== KET QUA DANH GIA SO BO =====")
print("Anh validation:", len(val_ds))
print("Anh co UAV duoc danh gia:", evaluated)

if ious:
    print(f"Mean IoU: {sum(ious)/len(ious):.4f}")
    print(f"Objectness trung binh: {sum(scores)/len(scores):.4f}")
    print("Luu y: day la IoU voi box dau tien, chua phai mAP.")
else:
    print("Khong tim thay box UAV trong validation.")

print("\nHOAN TAT KIEM TRA - KHONG HUAN LUYEN LAI")
```

```python
from pathlib import Path
from collections import Counter
import xml.etree.ElementTree as ET
from PIL import Image

data_dir = Path("/content/UAV_dataset_unar/dataset_xml_format")

images = {
    p.stem: p
    for p in data_dir.rglob("*")
    if p.is_file() and p.suffix.lower() in {".png", ".jpg", ".jpeg"}
}
xmls = {p.stem: p for p in data_dir.rglob("*.xml")}

print("Số ảnh:", len(images))
print("Số XML:", len(xmls))
print("Ảnh thiếu XML:", len(set(images) - set(xmls)))
print("XML thiếu ảnh:", len(set(xmls) - set(images)))

bad_images = []
bad_xml = []
bad_boxes = []
classes = Counter()
object_count = 0

for stem in sorted(set(images) & set(xmls)):
    try:
        with Image.open(images[stem]) as im:
            width, height = im.size
            im.verify()
    except Exception as e:
        bad_images.append((stem, str(e)))
        continue

    try:
        root = ET.parse(xmls[stem]).getroot()
        size = root.find("size")
        if size is None:
            raise ValueError("Thiếu thẻ size")

        ann_w = int(size.findtext("width", "0"))
        ann_h = int(size.findtext("height", "0"))
        if ann_w <= 0 or ann_h <= 0:
            raise ValueError("Kích thước XML không hợp lệ")

        for obj in root.findall("object"):
            name = (obj.findtext("name") or "").strip().lower()
            classes[name] += 1
            object_count += 1

            box = obj.find("bndbox")
            if box is None:
                bad_boxes.append((stem, "Thiếu bndbox"))
                continue

            xmin = float(box.findtext("xmin", "nan"))
            ymin = float(box.findtext("ymin", "nan"))
            xmax = float(box.findtext("xmax", "nan"))
            ymax = float(box.findtext("ymax", "nan"))

            if not (0 <= xmin < xmax <= ann_w and
                    0 <= ymin < ymax <= ann_h):
                bad_boxes.append((stem, (xmin, ymin, xmax, ymax,
                                         ann_w, ann_h)))
    except Exception as e:
        bad_xml.append((stem, str(e)))

print("\nẢnh lỗi đọc:", len(bad_images))
print("XML lỗi cấu trúc:", len(bad_xml))
print("Bounding box lỗi:", len(bad_boxes))
print("Tổng đối tượng:", object_count)
print("Phân bố nhãn:", classes)
print("\nVí dụ lỗi XML:", bad_xml[:5])
print("Ví dụ bounding box lỗi:", bad_boxes[:5])
```

```python

from pathlib import Path
import xml.etree.ElementTree as ET
import cv2
import csv
from collections import Counter

DATA_DIR = Path("/content/UAV_dataset_unar/dataset_xml_format")
OUT_CSV = Path("/content/dataset_quality_report.csv")
IMAGE_EXTS = {".jpg", ".jpeg", ".png"}

images = {}
for p in DATA_DIR.rglob("*"):
    if p.is_file() and p.suffix.lower() in IMAGE_EXTS:
        images[p.stem.lower()] = p

xmls = {
    p.stem.lower(): p
    for p in DATA_DIR.rglob("*.xml")
}

rows = []
class_counts = Counter()
issues = Counter()
usable_pairs = []

for stem, xp in sorted(xmls.items()):
    ip = images.get(stem)

    if ip is None:
        rows.append([stem, "", str(xp), "missing_image", 0, ""])
        issues["missing_image"] += 1
        continue

    # 1. Kiểm tra ảnh thực tế đọc được và kích thước
    img = cv2.imread(str(ip))
    if img is None:
        rows.append([stem, str(ip), str(xp), "unreadable_image", 0, ""])
        issues["unreadable_image"] += 1
        continue

    actual_h, actual_w = img.shape[:2]

    # 2. Kiểm tra XML
    if xp.stat().st_size == 0:
        rows.append([stem, str(ip), str(xp), "empty_xml", 0, ""])
        issues["empty_xml"] += 1
        continue

    try:
        root = ET.parse(xp).getroot()
    except Exception as e:
        rows.append([stem, str(ip), str(xp), "xml_parse_error", 0, str(e)])
        issues["xml_parse_error"] += 1
        continue

    # 3. So sánh kích thước khai báo trong XML với ảnh
    size = root.find("size")
    size_ok = True
    if size is not None:
        try:
            xml_w = int(size.findtext("width", "0"))
            xml_h = int(size.findtext("height", "0"))
            size_ok = (xml_w == actual_w and xml_h == actual_h)
        except (TypeError, ValueError):
            size_ok = False
    else:
        size_ok = False

    objects = root.findall("object")
    box_errors = []
    valid_objects = 0

    for obj in objects:
        cls = (obj.findtext("name") or "").strip()
        if cls:
            class_counts[cls] += 1

        bnd = obj.find("bndbox")
        if bnd is None:
            box_errors.append("missing_bndbox")
            continue

        try:
            xmin = float(bnd.findtext("xmin"))
            ymin = float(bnd.findtext("ymin"))
            xmax = float(bnd.findtext("xmax"))
            ymax = float(bnd.findtext("ymax"))
        except (TypeError, ValueError):
            box_errors.append("invalid_coordinates")
            continue

        if not (xmin < xmax and ymin < ymax):
            box_errors.append("invalid_box_order")
            continue

        # VOC thường dùng tọa độ pixel; cho phép tọa độ bắt đầu từ 0 hoặc 1.
        if (xmin < 0 or ymin < 0 or
            xmax > actual_w or ymax > actual_h):
            box_errors.append("box_out_of_image")
            continue

        valid_objects += 1

    status_parts = []
    if not size_ok:
        status_parts.append("size_mismatch_or_missing")
    if box_errors:
        status_parts.append("bbox_error")
    if len(objects) == 0:
        status_parts.append("no_objects")

    # Chỉ đưa vào danh sách ứng viên khi kích thước và mọi box đều hợp lệ,
    # đồng thời có ít nhất một object được gán nhãn.
    status = "candidate"
    if not size_ok or box_errors or len(objects) == 0:
        status = "|".join(status_parts) or "needs_review"
    else:
        usable_pairs.append((ip, xp))

    rows.append([
        stem, str(ip), str(xp), status,
        len(objects),
        f"image={actual_w}x{actual_h}; "
        f"xml_size_ok={size_ok}; valid_objects={valid_objects}; "
        f"box_errors={','.join(box_errors)}"
    ])
    issues[status] += 1

with OUT_CSV.open("w", newline="", encoding="utf-8-sig") as f:
    writer = csv.writer(f)
    writer.writerow([
        "stem", "image_path", "xml_path",
        "status", "object_count", "details"
    ])
    writer.writerows(rows)

print("===== KIỂM TRA CHẤT LƯỢNG =====")
print("Tổng ảnh:", len(images))
print("Tổng XML:", len(xmls))
print("Cặp ứng viên hợp lệ:", len(usable_pairs))
print("Số bản ghi báo cáo:", len(rows))
print("Phân loại:")
for k, v in issues.items():
    print(f"  {k}: {v}")

print("Số object theo lớp:", dict(class_counts))
print("CSV:", OUT_CSV)
```

```python

from pathlib import Path
import csv
import cv2
import xml.etree.ElementTree as ET

BASE = Path("/content/UAV_CNN_Grayscale96")
MANIFEST = BASE / "manifest.csv"
OUT = BASE / "processed"
IMG_SIZE = 96

# Pascal VOC: xmin, ymin, xmax, ymax
# Nhãn đầu ra được chuẩn hóa về [0, 1].
def read_voc_boxes(xml_path, width, height):
    root = ET.parse(xml_path).getroot()
    boxes = []

    for obj in root.findall("object"):
        name = (obj.findtext("name") or "").strip()
        if name != "drone":
            continue

        bnd = obj.find("bndbox")
        if bnd is None:
            raise ValueError(f"Thiếu bndbox: {xml_path}")

        xmin = float(bnd.findtext("xmin"))
        ymin = float(bnd.findtext("ymin"))
        xmax = float(bnd.findtext("xmax"))
        ymax = float(bnd.findtext("ymax"))

        # Chặn sai số nhỏ ở biên; từ chối box sai cấu trúc.
        if xmin >= xmax or ymin >= ymax:
            raise ValueError(f"Bounding box không hợp lệ: {xml_path}")

        xmin = max(0.0, min(xmin, width))
        xmax = max(0.0, min(xmax, width))
        ymin = max(0.0, min(ymin, height))
        ymax = max(0.0, min(ymax, height))

        if xmin >= xmax or ymin >= ymax:
            raise ValueError(f"Bounding box nằm ngoài ảnh: {xml_path}")

        boxes.append([
            xmin / width,
            ymin / height,
            xmax / width,
            ymax / height,
        ])

    return boxes

with MANIFEST.open("r", encoding="utf-8-sig", newline="") as f:
    rows = list(csv.DictReader(f))

summary = {}
processed_rows = []
errors = []

for row in rows:
    image_path = Path(row["image_path"])
    xml_path = Path(row["xml_path"])
    split = row["split"]

    try:
        image = cv2.imread(str(image_path), cv2.IMREAD_GRAYSCALE)
        if image is None:
            raise ValueError("Không đọc được ảnh grayscale")

        h, w = image.shape
        boxes = read_voc_boxes(xml_path, w, h)

        # Resize trực tiếp về 96x96; box chuẩn hóa không đổi.
        resized = cv2.resize(
            image, (IMG_SIZE, IMG_SIZE),
            interpolation=cv2.INTER_AREA
        )

        image_out_dir = OUT / split / "images"
        label_out_dir = OUT / split / "labels"
        image_out_dir.mkdir(parents=True, exist_ok=True)
        label_out_dir.mkdir(parents=True, exist_ok=True)

        # Tên dựa trên đường dẫn đầy đủ để tránh trùng stem.
        import hashlib
        key = hashlib.sha256(
            str(image_path).encode("utf-8")
        ).hexdigest()[:16]

        out_image = image_out_dir / f"{key}.png"
        out_label = label_out_dir / f"{key}.txt"

        if not cv2.imwrite(str(out_image), resized):
            raise IOError(f"Không ghi được ảnh: {out_image}")

        with out_label.open("w", encoding="utf-8") as f:
            for x1, y1, x2, y2 in boxes:
                f.write(
                    f"drone {x1:.8f} {y1:.8f} "
                    f"{x2:.8f} {y2:.8f}\n"
                )

        processed_rows.append({
            "key": key,
            "split": split,
            "image_path": str(out_image),
            "label_path": str(out_label),
            "width_original": w,
            "height_original": h,
            "object_count": len(boxes),
        })

        summary.setdefault(split, {"images": 0, "objects": 0})
        summary[split]["images"] += 1
        summary[split]["objects"] += len(boxes)

    except Exception as e:
        errors.append((str(image_path), str(xml_path), str(e)))

out_manifest = OUT / "processed_manifest.csv"
with out_manifest.open("w", encoding="utf-8-sig", newline="") as f:
    writer = csv.DictWriter(
        f,
        fieldnames=[
            "key", "split", "image_path", "label_path",
            "width_original", "height_original", "object_count"
        ]
    )
    writer.writeheader()
    writer.writerows(processed_rows)

print("===== KẾT QUẢ TIỀN XỬ LÝ =====")
for split in ("train", "val", "test"):
    print(split, summary.get(split, {"images": 0, "objects": 0}))
print("Đã xử lý:", len(processed_rows), "/", len(rows))
print("Lỗi:", len(errors))
print("Manifest:", out_manifest)

if errors:
    print("\nVí dụ lỗi:")
    for e in errors[:10]:
        print(e)
```

```python

import csv
from pathlib import Path

import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader
import cv2
import numpy as np

BASE = Path("/content/UAV_CNN_Grayscale96/processed")
MANIFEST = BASE / "processed_manifest.csv"

IMG_SIZE = 96
GRID_SIZE = 12
NUM_SLOTS = 2
VALUES_PER_SLOT = 5
BATCH_SIZE = 32
SEED = 42

torch.manual_seed(SEED)
np.random.seed(SEED)

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Device:", device)

class UAVGridDataset(Dataset):
    def __init__(self, manifest_path, split):
        with open(manifest_path, encoding="utf-8-sig", newline="") as f:
            all_rows = list(csv.DictReader(f))
        self.rows = [r for r in all_rows if r["split"] == split]

    def __len__(self):
        return len(self.rows)

    def __getitem__(self, idx):
        row = self.rows[idx]

        image = cv2.imread(
            row["image_path"], cv2.IMREAD_GRAYSCALE
        )
        if image is None or image.shape != (IMG_SIZE, IMG_SIZE):
            raise ValueError(f"Ảnh không hợp lệ: {row['image_path']}")

        # [1, 96, 96], giá trị pixel trong [0, 1].
        image = torch.from_numpy(
            image.astype(np.float32) / 255.0
        ).unsqueeze(0)

        # [12, 12, 2, 5]
        target = torch.zeros(
            GRID_SIZE, GRID_SIZE, NUM_SLOTS, VALUES_PER_SLOT,
            dtype=torch.float32
        )

        label_path = Path(row["label_path"])
        boxes = []

        for line in label_path.read_text(encoding="utf-8").splitlines():
            if not line.strip():
                continue

            parts = line.split()
            if len(parts) != 5 or parts[0] != "drone":
                raise ValueError(f"Nhãn không hợp lệ: {line}")

            x1, y1, x2, y2 = map(float, parts[1:])
            cx = (x1 + x2) / 2.0
            cy = (y1 + y2) / 2.0
            bw = x2 - x1
            bh = y2 - y1

            gx = min(GRID_SIZE - 1, int(cx * GRID_SIZE))
            gy = min(GRID_SIZE - 1, int(cy * GRID_SIZE))

            # Tâm tương đối bên trong ô lưới.
            local_x = cx * GRID_SIZE - gx
            local_y = cy * GRID_SIZE - gy
            boxes.append((gy, gx, bw * bh, local_x, local_y, bw, bh))

        # Gán box có diện tích lớn hơn vào slot 0 trước.
        # Dataset đã kiểm tra có tối đa 2 object/ô.
        grouped = {}
        for box in boxes:
            gy, gx = box[0], box[1]
            grouped.setdefault((gy, gx), []).append(box)

        for (gy, gx), cell_boxes in grouped.items():
            if len(cell_boxes) > NUM_SLOTS:
                raise ValueError(
                    f"Ô ({gy},{gx}) có {len(cell_boxes)} object, "
                    f"vượt NUM_SLOTS={NUM_SLOTS}: {label_path}"
                )

            cell_boxes.sort(key=lambda b: b[2], reverse=True)

            for slot, box in enumerate(cell_boxes):
                _, _, _, local_x, local_y, bw, bh = box
                target[gy, gx, slot] = torch.tensor(
                    [1.0, local_x, local_y, bw, bh]
                )

        return image, target


class SmallUAVCNN(nn.Module):
    def __init__(self):
        super().__init__()

        self.backbone = nn.Sequential(
            nn.Conv2d(1, 16, 3, padding=1, bias=False),
            nn.BatchNorm2d(16),
            nn.ReLU(inplace=True),
            nn.MaxPool2d(2),       # 96 -> 48

            nn.Conv2d(16, 32, 3, padding=1, bias=False),
            nn.BatchNorm2d(32),
            nn.ReLU(inplace=True),
            nn.MaxPool2d(2),       # 48 -> 24

            nn.Conv2d(32, 64, 3, padding=1, bias=False),
            nn.BatchNorm2d(64),
            nn.ReLU(inplace=True),
            nn.MaxPool2d(2),       # 24 -> 12

            nn.Conv2d(64, 64, 3, padding=1, bias=False),
            nn.BatchNorm2d(64),
            nn.ReLU(inplace=True),
        )

        self.head = nn.Conv2d(
            64, NUM_SLOTS * VALUES_PER_SLOT, kernel_size=1
        )

    def forward(self, x):
        x = self.backbone(x)           # [B, 64, 12, 12]
        x = self.head(x)               # [B, 10, 12, 12]
        x = x.permute(0, 2, 3, 1)      # [B, 12, 12, 10]
        return x.reshape(
            x.shape[0], GRID_SIZE, GRID_SIZE, NUM_SLOTS, VALUES_PER_SLOT
        )


train_ds = UAVGridDataset(MANIFEST, "train")
val_ds = UAVGridDataset(MANIFEST, "val")
test_ds = UAVGridDataset(MANIFEST, "test")

train_loader = DataLoader(
    train_ds, batch_size=BATCH_SIZE, shuffle=True,
    num_workers=0, generator=torch.Generator().manual_seed(SEED)
)
val_loader = DataLoader(val_ds, batch_size=BATCH_SIZE, shuffle=False)
test_loader = DataLoader(test_ds, batch_size=BATCH_SIZE, shuffle=False)

model = SmallUAVCNN().to(device)

images, targets = next(iter(train_loader))
images = images.to(device)
targets = targets.to(device)

with torch.no_grad():
    predictions = model(images)

print("Train/val/test:", len(train_ds), len(val_ds), len(test_ds))
print("Image batch:", tuple(images.shape))
print("Target batch:", tuple(targets.shape))
print("Prediction batch:", tuple(predictions.shape))
print("Parameters:", sum(p.numel() for p in model.parameters()))
print("Output shape đúng:", predictions.shape[1:] == (12, 12, 2, 5))
```

```python

import torch
import torch.nn.functional as F
from pathlib import Path

EPOCHS = 30
LR = 1e-3
WEIGHT_DECAY = 1e-4
LAMBDA_BOX = 5.0
LAMBDA_OBJ = 1.0
POS_WEIGHT = 10.0

OUT_DIR = Path("/content/UAV_CNN_Grayscale96/checkpoints")
OUT_DIR.mkdir(parents=True, exist_ok=True)
BEST_PATH = OUT_DIR / "cnn_baseline_best.pth"

optimizer = torch.optim.AdamW(
    model.parameters(), lr=LR, weight_decay=WEIGHT_DECAY
)
scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(
    optimizer, mode="min", factor=0.5, patience=3
)

def detection_loss(pred, target):
    # Objectness logits: chưa sigmoid.
    obj_logits = pred[..., 0]
    obj_target = target[..., 0]

    pos = obj_target > 0.5
    neg = ~pos

    # Tính BCE riêng cho positive và negative để kiểm soát mất cân bằng.
    pos_loss = (
        F.binary_cross_entropy_with_logits(
            obj_logits[pos], obj_target[pos], reduction="mean"
        )
        if pos.any() else obj_logits.sum() * 0.0
    )
    neg_loss = (
        F.binary_cross_entropy_with_logits(
            obj_logits[neg], obj_target[neg], reduction="mean"
        )
        if neg.any() else obj_logits.sum() * 0.0
    )
    obj_loss = POS_WEIGHT * pos_loss + neg_loss

    # Chỉ tính box loss ở slot có object.
    if pos.any():
        box_pred = pred[..., 1:5][pos]
        box_true = target[..., 1:5][pos]

        # local x/y nằm trong [0,1], w/h nằm trong [0,1].
        box_pred = torch.cat([
            box_pred[..., :2].sigmoid(),
            box_pred[..., 2:4].sigmoid()
        ], dim=-1)

        box_loss = F.smooth_l1_loss(box_pred, box_true)
    else:
        box_loss = pred.sum() * 0.0

    total = LAMBDA_OBJ * obj_loss + LAMBDA_BOX * box_loss
    return total, obj_loss.detach(), box_loss.detach()

@torch.no_grad()
def run_validation():
    model.eval()
    total_loss = 0.0
    total_obj = 0.0
    total_box = 0.0
    n = 0

    for x, y in val_loader:
        x, y = x.to(device), y.to(device)
        pred = model(x)
        loss, obj_loss, box_loss = detection_loss(pred, y)

        batch_n = x.size(0)
        total_loss += loss.item() * batch_n
        total_obj += obj_loss.item() * batch_n
        total_box += box_loss.item() * batch_n
        n += batch_n

    return total_loss / n, total_obj / n, total_box / n

best_val = float("inf")
history = []

for epoch in range(1, EPOCHS + 1):
    model.train()
    train_sum = 0.0
    train_n = 0

    for x, y in train_loader:
        x, y = x.to(device), y.to(device)

        optimizer.zero_grad(set_to_none=True)
        pred = model(x)
        loss, _, _ = detection_loss(pred, y)
        loss.backward()
        torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=5.0)
        optimizer.step()

        train_sum += loss.item() * x.size(0)
        train_n += x.size(0)

    train_loss = train_sum / train_n
    val_loss, val_obj, val_box = run_validation()
    scheduler.step(val_loss)

    history.append({
        "epoch": epoch,
        "train_loss": train_loss,
        "val_loss": val_loss,
        "val_obj_loss": val_obj,
        "val_box_loss": val_box,
    })

    print(
        f"Epoch {epoch:02d}/{EPOCHS} | "
        f"train={train_loss:.4f} | val={val_loss:.4f} | "
        f"val_obj={val_obj:.4f} | val_box={val_box:.4f}"
    )

    if val_loss < best_val:
        best_val = val_loss
        torch.save({
            "epoch": epoch,
            "model_state_dict": model.state_dict(),
            "optimizer_state_dict": optimizer.state_dict(),
            "best_val_loss": best_val,
            "config": {
                "img_size": IMG_SIZE,
                "grid_size": GRID_SIZE,
                "num_slots": NUM_SLOTS,
                "values_per_slot": VALUES_PER_SLOT,
                "seed": SEED,
            }
        }, BEST_PATH)
        print("  -> Saved best checkpoint:", BEST_PATH)

print("Best validation loss:", best_val)
print("Checkpoint:", BEST_PATH)
```

```python

import torch
import numpy as np
from pathlib import Path

CKPT = Path(
    "/content/UAV_CNN_Grayscale96/checkpoints/cnn_baseline_best.pth"
)

checkpoint = torch.load(CKPT, map_location=device, weights_only=False)
model.load_state_dict(checkpoint["model_state_dict"])
model.eval()

print("Checkpoint epoch:", checkpoint["epoch"])
print("Best validation loss:", checkpoint["best_val_loss"])

def box_iou_one_to_many(box, boxes):
    # xyxy, tọa độ chuẩn hóa
    if len(boxes) == 0:
        return np.zeros(0, dtype=np.float32)

    ix1 = np.maximum(box[0], boxes[:, 0])
    iy1 = np.maximum(box[1], boxes[:, 1])
    ix2 = np.minimum(box[2], boxes[:, 2])
    iy2 = np.minimum(box[3], boxes[:, 3])

    inter = np.maximum(0, ix2 - ix1) * np.maximum(0, iy2 - iy1)
    area1 = max(0, box[2] - box[0]) * max(0, box[3] - box[1])
    area2 = (
        np.maximum(0, boxes[:, 2] - boxes[:, 0]) *
        np.maximum(0, boxes[:, 3] - boxes[:, 1])
    )
    return inter / np.maximum(area1 + area2 - inter, 1e-9)

def nms(boxes, scores, iou_threshold=0.5):
    order = np.argsort(scores)[::-1]
    keep = []

    while len(order):
        i = order[0]
        keep.append(i)
        if len(order) == 1:
            break

        rest = order[1:]
        ious = box_iou_one_to_many(boxes[i], boxes[rest])
        order = rest[ious < iou_threshold]

    return np.asarray(keep, dtype=int)

@torch.no_grad()
def decode_batch(pred, conf_threshold=0.01, nms_iou=0.5):
    # pred: [B, 12, 12, 2, 5]
    pred = pred.detach().cpu()
    decoded = []

    for b in range(pred.shape[0]):
        boxes, scores = [], []

        for gy in range(GRID_SIZE):
            for gx in range(GRID_SIZE):
                for slot in range(NUM_SLOTS):
                    v = pred[b, gy, gx, slot]

                    score = torch.sigmoid(v[0]).item()
                    if score < conf_threshold:
                        continue

                    local_x, local_y = torch.sigmoid(v[1]).item(), torch.sigmoid(v[2]).item()
                    bw, bh = torch.sigmoid(v[3]).item(), torch.sigmoid(v[4]).item()

                    cx = (gx + local_x) / GRID_SIZE
                    cy = (gy + local_y) / GRID_SIZE
                    boxes.append([
                        max(0, cx - bw / 2),
                        max(0, cy - bh / 2),
                        min(1, cx + bw / 2),
                        min(1, cy + bh / 2),
                    ])
                    scores.append(score)

        if boxes:
            boxes = np.asarray(boxes, dtype=np.float32)
            scores = np.asarray(scores, dtype=np.float32)
            keep = nms(boxes, scores, nms_iou)
            decoded.append((boxes[keep], scores[keep]))
        else:
            decoded.append((
                np.zeros((0, 4), dtype=np.float32),
                np.zeros((0,), dtype=np.float32)
            ))

    return decoded

def get_gt_boxes(target):
    # target: [B, 12, 12, 2, 5]
    batch_gt = []

    for b in range(target.shape[0]):
        boxes = []
        for gy in range(GRID_SIZE):
            for gx in range(GRID_SIZE):
                for slot in range(NUM_SLOTS):
                    v = target[b, gy, gx, slot]
                    if v[0].item() < 0.5:
                        continue

                    cx = (gx + v[1].item()) / GRID_SIZE
                    cy = (gy + v[2].item()) / GRID_SIZE
                    bw, bh = v[3].item(), v[4].item()

                    boxes.append([
                        max(0, cx - bw / 2),
                        max(0, cy - bh / 2),
                        min(1, cx + bw / 2),
                        min(1, cy + bh / 2),
                    ])

        batch_gt.append(np.asarray(boxes, dtype=np.float32).reshape(-1, 4))

    return batch_gt

def average_precision_101(recalls, precisions):
    # COCO-style 101-point interpolation for one class and one IoU threshold.
    vals = []
    for r in np.linspace(0, 1, 101):
        p = precisions[recalls >= r]
        vals.append(float(p.max()) if len(p) else 0.0)
    return float(np.mean(vals))

def evaluate_ap(iou_threshold, all_preds, all_gt):
    total_gt = sum(len(g) for g in all_gt)
    if total_gt == 0:
        return 0.0, 0.0, 0.0

    flat = []
    for image_idx, (boxes, scores) in enumerate(all_preds):
        for j, score in enumerate(scores):
            flat.append((float(score), image_idx, boxes[j]))

    flat.sort(key=lambda x: x[0], reverse=True)
    matched = [np.zeros(len(g), dtype=bool) for g in all_gt]
    tp, fp = [], []

    for score, image_idx, box in flat:
        gt = all_gt[image_idx]
        if len(gt) == 0:
            tp.append(0)
            fp.append(1)
            continue

        ious = box_iou_one_to_many(box, gt)
        best = int(np.argmax(ious))

        if ious[best] >= iou_threshold and not matched[image_idx][best]:
            tp.append(1)
            fp.append(0)
            matched[image_idx][best] = True
        else:
            tp.append(0)
            fp.append(1)

    tp = np.cumsum(tp)
    fp = np.cumsum(fp)
    recalls = tp / max(total_gt, 1)
    precisions = tp / np.maximum(tp + fp, 1e-9)

    ap = average_precision_101(recalls, precisions)
    return ap, recalls[-1] if len(recalls) else 0.0, (
        precisions[-1] if len(precisions) else 0.0
    )

all_preds, all_gt = [], []

for x, y in val_loader:
    x = x.to(device)
    pred = model(x)
    all_preds.extend(decode_batch(pred, conf_threshold=0.01, nms_iou=0.5))
    all_gt.extend(get_gt_boxes(y))

ap50, recall_final, precision_final = evaluate_ap(0.50, all_preds, all_gt)
aps = [
    evaluate_ap(t, all_preds, all_gt)[0]
    for t in np.arange(0.50, 0.96, 0.05)
]

print("Validation ảnh:", len(all_gt))
print("Ground-truth boxes:", sum(len(g) for g in all_gt))
print(f"AP@0.50: {ap50:.4f}")
print(f"AP@0.50:0.95: {np.mean(aps):.4f}")
print(f"Precision cuối danh sách: {precision_final:.4f}")
print(f"Recall cuối danh sách: {recall_final:.4f}")
```

```python

import numpy as np

def metrics_at_threshold(all_preds, all_gt, conf_threshold, iou_threshold=0.5):
    tp = fp = 0
    total_gt = sum(len(g) for g in all_gt)

    for (boxes, scores), gt in zip(all_preds, all_gt):
        keep = np.where(scores >= conf_threshold)[0]
        boxes = boxes[keep]
        scores = scores[keep]

        order = np.argsort(scores)[::-1]
        matched = np.zeros(len(gt), dtype=bool)

        for idx in order:
            if len(gt) == 0:
                fp += 1
                continue

            ious = box_iou_one_to_many(boxes[idx], gt)
            best = int(np.argmax(ious))

            if ious[best] >= iou_threshold and not matched[best]:
                tp += 1
                matched[best] = True
            else:
                fp += 1

    fn = total_gt - tp
    precision = tp / (tp + fp) if tp + fp else 0.0
    recall = tp / total_gt if total_gt else 0.0
    f1 = (
        2 * precision * recall / (precision + recall)
        if precision + recall else 0.0
    )
    return tp, fp, fn, precision, recall, f1

print("Threshold | TP | FP | FN | Precision | Recall | F1")
for threshold in [0.01, 0.05, 0.10, 0.20, 0.30, 0.50, 0.70]:
    tp, fp, fn, p, r, f1 = metrics_at_threshold(
        all_preds, all_gt, threshold
    )
    print(
        f"{threshold:9.2f} | {tp:2d} | {fp:3d} | {fn:2d} | "
        f"{p:.4f} | {r:.4f} | {f1:.4f}"
    )
```

```python

import torch
import torch.nn.functional as F
from pathlib import Path

# Giữ nguyên model, train_loader, val_loader và device đã tạo.
# Huấn luyện lại từ đầu để so sánh tác động của HNM.
torch.manual_seed(42)
model_hnm = SmallUAVCNN().to(device)

EPOCHS_HNM = 30
LR_HNM = 3e-4
NEG_POS_RATIO = 3
LAMBDA_BOX = 5.0

def hnm_loss(pred, target):
    logits = pred[..., 0]
    obj = target[..., 0]
    positive = obj > 0.5
    negative = ~positive

    # BCE riêng từng slot, chưa giảm trung bình.
    per_slot = F.binary_cross_entropy_with_logits(
        logits, obj, reduction="none"
    )

    pos_loss = (
        per_slot[positive].mean()
        if positive.any() else logits.sum() * 0.0
    )

    # Chỉ giữ các negative có loss lớn nhất.
    neg_values = per_slot[negative]
    n_pos = int(positive.sum().item())
    n_keep = min(
        int(neg_values.numel()),
        max(1, NEG_POS_RATIO * max(n_pos, 1))
    )

    neg_loss = (
        torch.topk(neg_values, k=n_keep).values.mean()
        if n_keep > 0 else logits.sum() * 0.0
    )

    if positive.any():
        box_pred = pred[..., 1:5][positive]
        box_true = target[..., 1:5][positive]

        box_pred = torch.sigmoid(box_pred)
        box_loss = F.smooth_l1_loss(box_pred, box_true)
    else:
        box_loss = pred.sum() * 0.0

    total = pos_loss + neg_loss + LAMBDA_BOX * box_loss
    return total, pos_loss.detach(), neg_loss.detach(), box_loss.detach()


optimizer_hnm = torch.optim.AdamW(
    model_hnm.parameters(),
    lr=LR_HNM,
    weight_decay=5e-4
)
scheduler_hnm = torch.optim.lr_scheduler.ReduceLROnPlateau(
    optimizer_hnm, mode="min", factor=0.5, patience=3
)

hnm_dir = Path("/content/UAV_CNN_Grayscale96/checkpoints")
hnm_dir.mkdir(parents=True, exist_ok=True)
hnm_best_path = hnm_dir / "cnn_hnm_best.pth"

best_val_hnm = float("inf")
patience = 7
epochs_without_improvement = 0

@torch.no_grad()
def validate_hnm():
    model_hnm.eval()
    sums = [0.0, 0.0, 0.0, 0.0]
    n = 0

    for x, y in val_loader:
        x, y = x.to(device), y.to(device)
        pred = model_hnm(x)
        loss, pos, neg, box = hnm_loss(pred, y)

        bs = x.size(0)
        for i, value in enumerate([loss, pos, neg, box]):
            sums[i] += value.item() * bs
        n += bs

    return [v / n for v in sums]


for epoch in range(1, EPOCHS_HNM + 1):
    model_hnm.train()
    train_sum = 0.0
    n = 0

    for x, y in train_loader:
        x, y = x.to(device), y.to(device)

        optimizer_hnm.zero_grad(set_to_none=True)
        pred = model_hnm(x)
        loss, _, _, _ = hnm_loss(pred, y)
        loss.backward()
        torch.nn.utils.clip_grad_norm_(
            model_hnm.parameters(), max_norm=5.0
        )
        optimizer_hnm.step()

        train_sum += loss.item() * x.size(0)
        n += x.size(0)

    train_loss = train_sum / n
    val_loss, val_pos, val_neg, val_box = validate_hnm()
    scheduler_hnm.step(val_loss)

    print(
        f"Epoch {epoch:02d}/{EPOCHS_HNM} | "
        f"train={train_loss:.4f} | val={val_loss:.4f} | "
        f"pos={val_pos:.4f} | hard_neg={val_neg:.4f} | "
        f"box={val_box:.4f}"
    )

    if val_loss < best_val_hnm:
        best_val_hnm = val_loss
        epochs_without_improvement = 0
        torch.save({
            "epoch": epoch,
            "model_state_dict": model_hnm.state_dict(),
            "best_val_loss": best_val_hnm,
            "config": {
                "seed": 42,
                "loss": "hard_negative_mining",
                "negative_positive_ratio": NEG_POS_RATIO,
                "img_size": 96,
                "grid_size": 12,
                "num_slots": 2,
            }
        }, hnm_best_path)
        print("  -> Saved:", hnm_best_path)
    else:
        epochs_without_improvement += 1

    if epochs_without_improvement >= patience:
        print("Early stopping: validation loss không cải thiện.")
        break

print("Best HNM validation loss:", best_val_hnm)
print("Best checkpoint:", hnm_best_path)
```

```python

from pathlib import Path
import csv
import numpy as np
import torch

ROOT = Path("/content/UAV_CNN_Grayscale96")
CKPT_DIR = ROOT / "checkpoints"
MANIFEST = ROOT / "processed" / "processed_manifest.csv"

# Kiểm tra các biến và lớp được định nghĩa từ những ô trước.
required = [
    "UAVGridDataset", "SmallUAVCNN", "decode_batch",
    "get_gt_boxes", "box_iou_one_to_many",
    "average_precision_101", "GRID_SIZE", "NUM_SLOTS",
    "device"
]
missing = [name for name in required if name not in globals()]
if missing:
    raise RuntimeError(
        "Notebook chưa có các định nghĩa sau: "
        + ", ".join(missing)
        + ". Không cần tìm thủ công; gửi mình thông báo này."
    )

with open(MANIFEST, encoding="utf-8-sig", newline="") as f:
    rows = list(csv.DictReader(f))

val_ds = UAVGridDataset(MANIFEST, "val")
val_loader_check = torch.utils.data.DataLoader(
    val_ds, batch_size=32, shuffle=False, num_workers=0
)

def evaluate_ap_fixed(iou_threshold, all_preds, all_gt):
    total_gt = sum(len(g) for g in all_gt)
    if total_gt == 0:
        return 0.0

    flat = []
    for image_idx, (boxes, scores) in enumerate(all_preds):
        for j, score in enumerate(scores):
            flat.append((float(score), image_idx, boxes[j]))
    flat.sort(key=lambda item: item[0], reverse=True)

    matched = [np.zeros(len(g), dtype=bool) for g in all_gt]
    tp, fp = [], []

    for score, image_idx, box in flat:
        gt = all_gt[image_idx]
        unmatched = np.where(~matched[image_idx])[0]

        if len(unmatched) == 0:
            tp.append(0)
            fp.append(1)
            continue

        ious = box_iou_one_to_many(box, gt)
        best = unmatched[np.argmax(ious[unmatched])]

        if ious[best] >= iou_threshold:
            tp.append(1)
            fp.append(0)
            matched[image_idx][best] = True
        else:
            tp.append(0)
            fp.append(1)

    if not tp:
        return 0.0

    tp = np.cumsum(tp)
    fp = np.cumsum(fp)
    recalls = tp / total_gt
    precisions = tp / np.maximum(tp + fp, 1e-9)
    return average_precision_101(recalls, precisions)

print("Validation images:", len(val_ds))
print("Ground-truth boxes:", end=" ")

# Đọc GT một lần.
all_gt_check = []
for _, y in val_loader_check:
    all_gt_check.extend(get_gt_boxes(y))
print(sum(len(g) for g in all_gt_check))

for ckpt_name in ["cnn_baseline_best.pth", "cnn_hnm_best.pth"]:
    path = CKPT_DIR / ckpt_name
    if not path.exists():
        print(f"\n{ckpt_name}: không tìm thấy")
        continue

    checkpoint = torch.load(
        path, map_location=device, weights_only=False
    )
    candidate = SmallUAVCNN().to(device)
    candidate.load_state_dict(checkpoint["model_state_dict"])
    candidate.eval()

    preds = []
    with torch.no_grad():
        for x, _ in val_loader_check:
            x = x.to(device)
            pred = candidate(x)
            preds.extend(
                decode_batch(
                    pred, conf_threshold=0.01, nms_iou=0.5
                )
            )

    ap50 = evaluate_ap_fixed(0.50, preds, all_gt_check)
    ap_values = [
        evaluate_ap_fixed(float(t), preds, all_gt_check)
        for t in np.arange(0.50, 0.96, 0.05)
    ]

    print(f"\n{ckpt_name}")
    print("Checkpoint epoch:", checkpoint.get("epoch", "không lưu"))
    print("Số dự đoán sau NMS:", sum(len(b) for b, s in preds))
    print(f"AP@0.50 (đã sửa ghép GT): {ap50:.4f}")
    print(f"AP@0.50:0.95 (đã sửa ghép GT): {np.mean(ap_values):.4f}")
```

```python

import numpy as np
import torch
from pathlib import Path

ROOT = Path("/content/UAV_CNN_Grayscale96")
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

# Dùng lại validation loader và các hàm đã có trong notebook.
if "val_loader_check" not in globals():
    raise RuntimeError("Chưa có val_loader_check. Gửi mình thông báo lỗi này.")

def match_at_threshold(preds, gt_list, conf, iou_threshold=0.5):
    tp = fp = fn = 0

    for (boxes, scores), gt in zip(preds, gt_list):
        keep = scores >= conf
        boxes = boxes[keep]

        used = np.zeros(len(gt), dtype=bool)

        # Dự đoán đang theo thứ tự confidence giảm dần sau NMS.
        for box in boxes:
            if len(gt) == 0:
                fp += 1
                continue

            ious = box_iou_one_to_many(box, gt)
            unmatched = np.where(~used)[0]

            if len(unmatched) == 0:
                fp += 1
                continue

            best = unmatched[np.argmax(ious[unmatched])]
            if ious[best] >= iou_threshold:
                tp += 1
                used[best] = True
            else:
                fp += 1

        fn += int((~used).sum())

    precision = tp / max(tp + fp, 1)
    recall = tp / max(tp + fn, 1)
    f1 = 2 * precision * recall / max(precision + recall, 1e-12)
    return tp, fp, fn, precision, recall, f1

ckpt_path = ROOT / "checkpoints" / "cnn_hnm_best.pth"
checkpoint = torch.load(ckpt_path, map_location=device, weights_only=False)

candidate = SmallUAVCNN().to(device)
candidate.load_state_dict(checkpoint["model_state_dict"])
candidate.eval()

raw_preds = []
with torch.no_grad():
    for x, _ in val_loader_check:
        pred = candidate(x.to(device))
        raw_preds.extend(
            decode_batch(pred, conf_threshold=0.001, nms_iou=0.5)
        )

print("Ngưỡng | TP | FP | FN | Precision | Recall | F1")
for conf in [0.10, 0.20, 0.30, 0.40, 0.50, 0.60, 0.70, 0.80, 0.90]:
    tp, fp, fn, p, r, f1 = match_at_threshold(
        raw_preds, all_gt_check, conf
    )
    print(
        f"{conf:.2f} | {tp:3d} | {fp:4d} | {fn:3d} | "
        f"{p:.4f} | {r:.4f} | {f1:.4f}"
    )
```
